### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">Loading an image</mark>

### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">Accessing an entire row or column of a `numpy` array</mark>
We can also access pixels across an entire row or column of a 2D image. 

#### Accessing an entire row
To access pixels across an entire row, we would specify the target row's index along axis 0, and then collect all columns across it. In Python, we can specify we want all columns across a row in a few ways: 
```python
target_row = 1 # let's grab the second row in the image
image[target_row, :] # use : for the axis 1 index to collect all columns
image[target_row] # not specify a second index at all; if only one index is supplied, it applies to axis 0
```

<p class="alert alert-info">
    <strong>Note:</strong> If only 1 index is provided for a <b> numpy </b> array, then Python applies that supplied index to axis 0 of the array. This is done regardless of how many dimensions the <b> numpy </b> array has.
</p>

#### Accessing an entire column
To access pixels across an entire column, we would specify the target column and then collect all rows across it. We can write that in a few different ways as well: 
```python
target_column = 3 # let's grab the fourth column in the image
image[:, target_column] # use : for the axis 0 index to collect all rows
```

## <mark style="color: black; background-color: rgb(127,196,125); padding: 3px; border-radius: 5px;">Multidimensional Images</mark>

### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">Concept</mark>
 
So far we have been discussing `numpy` arrays in the context of individual 2D images. However, `numpy` arrays can also be applied to *multidimensional images*. Common examples of multidimensional fluorescence images include z-stacks, time lapses, multiple fluorescence channels, etc. In each of these examples, an additional dimension is created to individual 2D images with x and y dimensions. 

### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">How a 3D image set is organized in a `numpy` array</mark>
Let's consider the following very simple 3D image set containing 3 identical images:

This 3D image set spans 3 dimensions: it contains 3 *images*, each with 3 *rows* of pixels and 3 *columns* of pixels. Just as with individual 2D images, a `numpy` array views these 3 dimensions as 3 *axes*, where the first axis is the image dimension (axis 0), the second axis is the row dimension of a given image (axis 1), and the third axis is the column dimension of a given image (axis 2).

Below is how this simple 3D image is represented as a `numpy` array
```python

       [[5, 1, 3],
```

### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">Image projections</mark>

Z-stacks are a common example of multidimensional images in fluorescence microscopy. Z-stacks contain 2D images collected at different z planes to sample a 3D volume. One way to simplify viewing these 3D images is to collapse the data down to 2D using **image projections**.

Below are examples of common image projections used in fluorescence microscopy and their corresponding `numpy` functions. You can refer to this `numpy` [documentation page](https://numpy.org/doc/2.2/reference/routines.statistics.html) for more details about the functions. 

| Type of Projection | Function | What it does along an axis | Example projection (`axis=0`)                                   |
| ------------------------ | --------------- | ---------------------------------- | --------------------------------------------------------------- |
| Maximum Intensity Projection | `np.max` | Takes the maximum value | `mip = np.max(stack, axis=0)` |
| Minimum Intensity Projection | `np.min` | Takes the minimum value | `min_proj = np.min(stack, axis=0)` |
| Sum Projection | `np.sum` | Takes the sum | `sum_proj = np.sum(stack, axis=0)` |
| Average Intensity Projection | `np.mean` | Takes the arithmetic mean | `mean_proj = np.mean(stack, axis=0)` |
| Standard Deviation Projection | `np.std` | Takes the standard deviation | `sigma_map = np.std(stack, axis=0)` |
| Variance Projection | `np.var` | Takes the variance | `var_map = np.var(stack, axis=0)` |

## <mark style="color: black; background-color: rgb(127,196,125); padding: 3px; border-radius: 5px;">5. Saving Images</mark>

### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">Concept</mark>

With ```tifffile```, we can also save output files using the function `imwrite()`.

### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">Saving Output Files</mark>
To save files using `imwrite()`, we can specify two things:  
1. Required: The **file path** the image will be saved to, which ends with a specified **file name**
2. Optional: The **dtype** of the image to be saved; if not provided, `imwrite()` will infer the dtype from the array

Here's how we can write the code to save an image `output_image`: 
```python
import tifffile as tiff # this would be at the beginning of the code, and only specified once
output_path = "/Users/edelase/bobiac/results/outputimage.tif" # the full file path including the file name we want to save
tiff.imwrite(output_path, output_image.astype("uint8"))
```

Here, `tiff.imwrite` is provided with 2 inputs: 
1. Required: `output_path` is the **file path** the image will be saved to, which ends with a specified **file name**
2. Optional: `output_image.astype("uint8")` is a specification for the data being saved as "uint8", or 8 bit

`tiff.imwrite()` will use these two inputs to output a file saved to the folder specified in `output_path`. 

<p class="alert alert-info">
    <strong>NOTE:</strong> The <b>dtype</b> of a labeled image is important because it determines the maximum number of intensity values stored in the image. This detail will become particularly important later in the course when we start doing mathematical operations with images.
</p>

So far we have been using `ndv` to view multidimensional images. We can also use `matplotlib`, but doing so would involve plotting multiple images. In practice, this is done by creating multiple subplots (1 for each image) within an overall figure.

### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">Creating a figure with subplots</mark>
Using `matplotlib`, we can create a figure with subplots by writing the following:
```python
# Create a figure with axes
fig, axes = plt.subplots(1, 2, figsize=(8, 6))  # 1 row, 2 columns
```

### <mark style="color: black; background-color: rgb(190,223,185); padding: 3px; border-radius: 5px;">Filling the subplots with images</mark>

```python
 # Plot images
```
